3. Fes una petició a una publicació inexistent per obtenir un error 404 i mostra el codi d'estat rebut.

```python
#Obtener una publicación inexistente y mostrar el código de estatus
url = "https://jsonplaceholder.typicode.com/posts/9999"

try:
    response = requests.get(url)
    
    print(f"Código de estado: {response.status_code}")
    
    if response.status_code == 200:
        print("GET request successful!")
        print("Response:")
        print(response.json())
    elif response.status_code == 404:
        print("Error 404: La publicación no existe")
        print("Response:")
        print(response.text)
    else:
        print(f"GET request failed with status code {response.status_code}")
        
except requests.exceptions.RequestException as e:
    print(f"Error al realizar la petición: {e}")
```

```python
#Realizar una petición GET - Using a public API that returns JSON
url = "https://api.thedogapi.com/v1/breeds"
response = requests.get(url)

# Mostrar el código de estado
print(f"Código de estado: {response.status_code}")

# Verificar si la respuesta es JSON antes de parsear
if response.status_code == 200:
	try:
		data = response.json()
		# Imprimir algunos campos de la respuesta
		print(f"\nTotal de razas de perros: {len(data)}")
		print("\nPrimeras 3 razas:")
		for breed in data[:3]:
			print(f"  - Nombre: {breed.get('name', 'N/A')}")
			print(f"    Origen: {breed.get('origin', 'N/A')}")
			print(f"    Temperamento: {breed.get('temperament', 'N/A')}\n")
	except requests.exceptions.JSONDecodeError:
		print("Error: La respuesta no es JSON válido")
		print(response.text[:200])
else:
	print(f"Error en la petición: {response.status_code}")
```

4. Converteix els resultats a un DataFrame.

```python
import pandas as pd


# URL base
url_base = 'https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search'

# Parámetros de consulta
parametros = {
    'resource_id': '3dce0dbd-5a7d-439a-b945-6bffc17a6c03',
    'limit': 100
}

# 1. Realizar el GET request
response = requests.get(url_base, params=parametros)
response.raise_for_status() # Buena práctica para lanzar una excepción si el request falla

# 2. Convertir la respuesta a JSON
data = response.json()

# 3. Extraer la lista de registros ('records')
# Los datos tabulares de CKAN suelen estar en data['result']['records']
try:
    registros = data['result']['records']
except KeyError:
    print("Error: No se encontró la clave 'result' o 'records' en la respuesta de la API.")
    registros = [] # Inicializa como lista vacía en caso de error

# 4. Crear el DataFrame
df = pd.DataFrame(registros)

# Mostrar las primeras filas del DataFrame (opcional)
df.head()
```
