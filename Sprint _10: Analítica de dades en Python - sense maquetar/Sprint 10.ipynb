```python
#Importar las librerías 'numpy' y 'matplotlib' para realizar las gráficas solicitadas
import numpy as np
import matplotlib.pyplot as plt

#Crear una función que clasifica las columnas del DataFrame y genera un gráfico apropiado para cada una según el topo de variable
# Numérica, Categórica, Fecha (Histograma, Barras de Frecuencia o Barras de Años).
#Args: El DataFrame de pandas a analizar y graficar.
def clasificar_y_graficar_variables(df: pd.DataFrame):
    
    # Identificar columnas numéricas
    columnas_num = df.select_dtypes(include=np.number).columns.tolist()
    
    # Identificar columnas de fecha/hora (datetime)
    # Usamos 'datetime64' para capturar todos los subtipos de fecha de NumPy/Pandas
    columnas_fecha = df.select_dtypes(include=['datetime', 'datetime64']).columns.tolist()

    # Inicializar el diccionario de resultados
    clasificacion = {
        'numericas': columnas_num,
        'fechas': columnas_fecha,
        # Si no es numérica ni fecha, se clasifica como categórica (object, bool, category, etc.)
        'categoricas': [col for col in df.columns if col not in columnas_num and col not in columnas_fecha]
    }
    #Imprimimos las varias clasificadas según tipo de dato
    print(clasificacion)
    
    
    # Gráficos para Columnas Numéricas (Histograma)
    # Bucle for para iterar sobre todas las columnas y clasificar los gráficos por tipo de dato numérico
    for col in clasificacion['numericas']:
        plt.figure(figsize=(12, 5))
        df[col].plot(kind='hist', bins=15, edgecolor='black', color='skyblue')
        plt.title(f'Histograma de Frecuencia de {col}', fontsize=12)
        plt.xlabel(col)
        plt.ylabel('Frecuencia')
    
        
        nombre_archivo_imagen = f"grafico_numerica_{col}.png"
        plt.savefig(nombre_archivo_imagen)
        plt.close()
        print(f"Gráfico guardado: {nombre_archivo_imagen}")


    # Gráficos para Columnas Categóricas (Barras de Frecuencia)
    for col in clasificacion['categoricas']:
        plt.figure(figsize=(8, 6))
        # Contar la frecuencia de los valores y graficar las barras
        conteo = df[col].value_counts().head(10) # Mostrar los 10 más frecuentes
        conteo.plot(kind='bar', color='green', alpha=0.7)
        
        plt.title(f'Frecuencia de las Categorías en {col}', fontsize=12)
        plt.xlabel(col)
        plt.ylabel('Conteo')
        plt.xticks(rotation=45, ha='right') # Rotar etiquetas para mejor lectura
        plt.tight_layout()
        
        nombre_archivo_imagen = f"grafico_categorica_{col}.png"
        plt.savefig(nombre_archivo_imagen)
        plt.close()
        print(f"Gráfico guardado: {nombre_archivo_imagen}")


    # Gráficos para Columnas de Fecha/Hora (Barras de Frecuencia por Año)
    for col in clasificacion['fechas']:
        plt.figure(figsize=(8, 6))
        # Extraer el año y contar su frecuencia
        conteo_anual = df[col].dt.year.value_counts().sort_index()
        conteo_anual.plot(kind='bar', color='purple', alpha=0.7)
        
        plt.title(f'Frecuencia Anual de {col}', fontsize=12)
        plt.xlabel('Año')
        plt.ylabel('Conteo de Registros')
        plt.xticks(rotation=0)
        plt.tight_layout()
        
        nombre_archivo_imagen = f"grafico_fecha_{col}.png"
        plt.savefig(nombre_archivo_imagen)
        plt.close()
        print(f"Gráfico guardado: {nombre_archivo_imagen}")
    
graficos = clasificar_y_graficar_variables(df)
```

2.
Carrega l'arxiu matriu_distancies.xlsx a pandas, de manera que els noms de files i els noms de columnes siguin els de les ciutats. Borra "Las Palmas de Gran Canaria" i "Palma" perquè poguem fer el trajecte en cotxe.

Ens interessa visitar totes les ciutats principals d'Espanya recorrent la mínima distància possible.

No cal que ho facis de forma òptima, ens interessa que desenvolupis una solució raonable utilitzant les eines que tens actualment.

Per exemple, una aproximació senzilla (que no òptima) seria anant sempre a la ciutat més propera que no haguem visitat encara

Fes una funció que donada la matriu de distàncies i la ciutat d'origen, faci una proposta de ruta que sigui el més curta possible que puguis, retornant una llista amb l'ordre de visita. Dóna també la distància total recorreguda.

(EXTRA) Des de quina ciutat la ruta seria més curta amb l'algoritme plantejat

```python
#Cargar el archivo matriz con las distancias entre ciudades
import pandas as pd
df = pd.read_excel(r"C:\Users\jprom\Desktop\IT Academy\Python\Data Analytics\Sprint 10\matriu_distancies.xlsx")
df_distancies = df.copy()

#Eliminar las columnas y filas de las ciudades : Palma', 'Las Palmas de Gran Canaria'.
df_distancies = df.drop(['Palma', 'Las Palmas de Gran Canaria'], axis=1, inplace=True)
df_distancies = df.drop([6,7])

#Renombrar la columna 'Unnamed' y luego asignar esta como índice del DF
df_distancies_idx = df_distancies.rename(columns={'Unnamed: 0': 'Ciudades'})
df_distancies_idx.set_index('Ciudades', inplace=True)
df_distancies_idx
```

```python
#Crear una función que toma como argumentos el nombre de una ciudad, y un DF con la matriz de distancias entre ciudades
# y retorne la ruta mas corta y la cantidad de kms recorridos

def viajar(ciudad_inicio, df):
    ciudad_actual = ciudad_inicio
    ciudades = list(df_distancies_idx.columns)
    #Convertir a un conjunto el nombre de las ciudades para poder iterar sobre ellas
    ciudades_por_visitar = set(ciudades)
    #Remmover del conjunto de ciudades, la indicada como ciudad de inicio para que no la contabilice en el bucle
    ciudades_por_visitar.remove(ciudad_actual)
    #Crear una lista con el nombre de la ciudad de inicio, para ir agregando el resto de ciudades de la ruta 
    ruta = [ciudad_actual]
    #Crear un contador a 0, para contabilizar las distancias en Kms
    distancia_total = 0
    #Crear un bucle while con la condición de que recorra todas las ciudades por visitar
    while ciudades_por_visitar:
        distancia_actual = df.loc[:,ciudad_actual]
        distancia_filtrada = distancia_actual[list(ciudades_por_visitar)]
        #Encontrar de las distancias filtradas el valor mínimo mas cercano
        try:
            ciudad_siguiente = distancia_filtrada.idxmin()
            distancia_recorrida = distancia_filtrada.min()
        except ValueError:
            # Esto debería ocurrir solo si quedan 0 ciudades_por_visitar
            break
        #Actualizar el listado de la ruta con la siguiente ciudad tras el primer bucle, y sumamos la distancia al contador
        ruta.append(ciudad_siguiente)
        distancia_total += distancia_recorrida
        #Remover del listado de ciudades por visitar, la que ya agregamos al listado de la ruta, para actualizar la lista
        ciudades_por_visitar.remove(ciudad_siguiente)
        #Actualizar la ciudad actual, con el nombre de la siguiente ciudad a recorrer por el bucle, por distancia más cercana
        ciudad_actual = ciudad_siguiente
    #Regresar a la ciudad de origen para cerrar el ciclo
    ciudad_origen = ruta[0]
    distancia_regreso = df.loc[ciudad_actual, ciudad_origen]
    ruta.append(ciudad_origen)
    distancia_total += distancia_regreso
    return ruta, distancia_total


#Input por el usuario de la ciudad de inicio de su viaje
origen = str(input('Ingrese la ciudad de origen de su ruta')).capitalize()

#Llamar a la función con los parametros requeridos
ruta_sugerida, distancia_total = viajar(origen, df_distancies_idx)

print(f"Partiendo desde", origen, "la mejor ruta para viajar es:", ruta_sugerida, "\n","La distancia total recorrida es de:", distancia_total, "kms")

data = {'inicio': [origen],'ruta': [ruta_sugerida],"distancia":[distancia_total]}
df_rutas = pd.DataFrame(data)
df_rutas
```

```python

# Crear una lista vacía, para alojar las posibles rutas
mejores_rutas = []

# Crear la variable ciudades_inicio con los nombres de las columnas del DF
ciudades_inicio = df_distancies_idx.columns

#Iterar con un bucle for sobre todas las ciudades de inicio posibles para encontrar la mejor ruta
for ciudad in ciudades_inicio:
    ruta, distancia = viajar(ciudad, df_distancies_idx)
    #Anexar en la lista mejores_rutas, las tres variables ciudad, ruta, distancia
    mejores_rutas.append({
        'inicio': ciudad,
        'ruta': ruta,
        'distancia': distancia
    })

#Encontrar la ruta con la menor distancia total
df_resultados = pd.DataFrame(mejores_rutas)
mejor_ruta = df_resultados.loc[df_resultados['distancia'].idxmin()]

# Mostrar el Resultado Final 
print(f"La ruta mas corta para visitar las principales ciudades de España, es partiendo desde", mejor_ruta['inicio'],"\n" "con una distancia total recorrida de:", mejor_ruta['distancia'], "kms")
df_resultados
```
