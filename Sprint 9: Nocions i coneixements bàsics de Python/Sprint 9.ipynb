1. Calculadora de l'índex de massa corporal

Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades.
La funció ha de classificar el resultat en les seves respectives categories.

Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduïdes per l’usuari/a no estan en el format adequat o no pren valors raonables.

```python
# Definimos la función
def calc_imc(peso, altura):
    return round(peso / (altura ** 2),2)
```

2. Convertidor de temperatures

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un print() no es pot guardar mai).

Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduides per l’usuari/a no estan en el format adequat.

(EXTRA): Pensa una manera d’emmagatzemar totes les possibles conversions en un sol objecte (Llista? Diccionari? DataFrame?) en comptes d’escriure molts if else en funció de la temperatura d’origen i la temperatura de destí.

```python
#Definimos las funciones para cada tipo de conversión de unidad de temperatura
def cel_a_kelvin(valor):
    return valor + 273.15

def cel_a_far(valor):
    return (valor * 9/5) + 32

def kelvin_a_cel(valor):
    return valor - 273.15

def kelvin_a_far(valor):
    return (valor - 273.15) * 9/5 + 32

def far_a_cel(valor):
    return (valor - 32) * 5/9

def far_a_kel(valor):
    return (valor - 32) * 5/9 + 273.15

# Si las unidades son las mismas, no se hace nada
def idem(valor):
    return valor
```

```python
#Creamos un diccionario con las funciones de conversión y un DataFrame que funcione como matriz de las operaciones

import pandas as pd

# Diccionario de funciones de conversión
operaciones = {
    'CelaK': cel_a_kelvin,
    'CelaF': cel_a_far,
    'KaCel': kelvin_a_cel,
    'KaFar': kelvin_a_far,
    'FaCel': far_a_cel,
    'FaKel': far_a_kel,
    'Idem': idem, # Para el caso de misma unidad
}

# DataFrame para mapear las operaciones
df = pd.DataFrame(
    [
        ['Idem', 'CelaK', 'CelaF'],
        ['KaCel', 'Idem', 'KaFar'],
        ['FaCel', 'FaKel', 'Idem']
    ],
    index=['CELSIUS', 'KELVIN', 'FAHRENHEIT'],
    columns=['CELSIUS', 'KELVIN', 'FAHRENHEIT']
)

df
```

```python
#Creamos la función de conversor con los argumentos de entrada

def conversor_temp(u_origen, u_destino, valor, df=df):
    """
    Convierte un valor de temperatura de una unidad a otra.

    Argumentos:
        u_origen (str): Unidad de temperatura de origen (ej. 'Celsius').
        u_destino (str): Unidad de temperatura de destino (ej. 'Kelvin').
        valor (float): El valor numérico a convertir.
        df (pd.DataFrame): DataFrame que mapea las operaciones.
    
    Return:
        float: El valor de la temperatura convertido.
    """
    # Normalizar las unidades a mayúsculas
    u_origen = u_origen.upper()
    u_destino = u_destino.upper()

    # Obtener el nombre de la operación del DataFrame
    nombre_operacion = df.loc[u_origen, u_destino]
    
    # Buscar la función de conversión en el diccionario 'operaciones'
    funcion_conversion = operaciones[nombre_operacion]
    
    # Ejecutar la función con el valor de entrada y devolver el resultado
    return funcion_conversion(float(valor))

temp_celsius = 25
temp_kelvin = conversor_temp('Celsius', 'Kelvin', temp_celsius)
print(f"La temperatura de {temp_celsius}°C es {temp_kelvin}K.")

temp_far = conversor_temp('Kelvin', 'Fahrenheit', temp_kelvin)
print(f"La temperatura de 298.15K es {temp_far}°F.")

temp_c_a_f = conversor_temp('Celsius', 'Fahrenheit', temp_far)
print(f"La temperatura de 100°C es {temp_c_a_f}°F.")

temp_kel_a_cel = conversor_temp('Kelvin', 'Celsius', temp_c_a_f)
print(f"La temperatura de 0K es {temp_kel_a_cel}°C.")
```

```python
#Creamos una lista vacía para almacenar las unidades y los valores de la conversión 
#Creamos un bucle para iterar por cada una de las posibles opciones

lista_medidas = []
finalizar = True

# Unidades válidas
unidades_validas = ['CELSIUS', 'KELVIN', 'FAHRENHEIT']

while finalizar:
    #Solicitamos los datos a convertir al usuario 
    unid1 = input("Ingrese la unidad de temperatura de origen: (ej. 'Celsius')").strip().upper()
    unid2 = input("Ingrese la unidad de la temperatura destino: (ej. 'Kelvin') ").strip().upper()
    temp = input("Ingrese el valor numérico a convertir: ")

    # Validación de datos
    if not unid1 or not unid2 or not temp:
        print("Debes ingresar el nombre correcto de la unidad de medida y el valor a convertir")
        continue
    if unid1 not in unidades_validas or unid2 not in unidades_validas:
        print(f"Unidades no válidas. Las opciones son: {', '.join(unidades_validas)}")
        continue
    try:
        temp_float = float(temp)
    except ValueError:
        print("El valor de temperatura debe ser numérico.")
        continue

    # Realizar la conversión
    operacion = conversor_temp(unid1, unid2, temp_float)
    lista_medidas.append({'origen': unid1, 'destino': unid2, 'valor': temp_float, 'resultado': operacion})

    continuar = input("¿Quiere convertir la temperatura a alguna otra unidad? (SI/NO): ")
    if continuar.strip().upper() == "NO":
        finalizar = False

lista_medidas 
```

Aquí también se ha utilizado el manejo de excepciones de error, mediante try/except, para posibles errores de valor, al input de los datos por parte del usuario. Tambíen mediante la función "strip" se eliminan caracteres no deseados en el input, y con "upper" convertimos todos los caracteres del "string" a mayúsculas.

```python
file=(r"C:\Users\jprom\Desktop\IT Academy\Python\Data Analytics\Sprint 9\tu me quieres blanca.txt")

#Creamos una función para almacenar en una lista todas las palabras del archivo de texto importado
def lista_palabras_archivo(filename):
    #Utilizamos el método with open para que el archivo importado se cierre, luego de su uso
    with open(filename, "r", encoding="utf-8") as f:
        palabras_extraidas = []
        for line in f:
            #Dividimos cada linea de la lista en los elementos/palabras que la componen y convertimos a minúsculas
            lista_palabras = line.lower().split()
            #Con la función extend fusionamos los diferentes elementos iterables de cada linea del texto, en una lista
            palabras_extraidas.extend(lista_palabras)
        return palabras_extraidas

lista_palabras_txt= lista_palabras_archivo(file)
print(lista_palabras_txt)
```

```python
#Creamos una nueva función para almacenar en un diccionario las palabras de la lista y cantidad, como key y value
def contar_palabras(lista):
    contador_palabras = {}
    for palabra in lista:
        # Limpia la palabra de comas, puntos, signos, etc. y la convierte a minúsculas
        palabra_limpia = palabra.replace(',', '').replace('.', '').replace('(', '').replace('¡','').lower()
        # Ignora las cadenas vacías que podrían resultar de la limpieza
        if palabra_limpia:
        # Incrementa el conteo de la palabra si ya existe, si no, la inicializa en 1
            if palabra_limpia in contador_palabras:
                contador_palabras[palabra_limpia] += 1
            else:
                contador_palabras[palabra_limpia] = 1
    return contador_palabras

# Llama a la función y muestra el resultado
resultado = contar_palabras(lista_palabras_txt)
print(resultado)
```

```python
#Creamos una función que toma como parámetro el diccionario con las claves y valores
def indice_palabras(diccionario):
    #Diccionario para alojar las claves como indices
    dic_final = {} 
    #Separamos la clave y valor del diccionario parámetro
    for palabra, cantidad in diccionario.items(): 
        primera_letra = palabra[0] 
        #Condicional para adicionar al diccionario final la primer letra de cada clave
        if primera_letra not in dic_final: 
            dic_final[primera_letra] = {}
        #Asignamos la primera letra y la palabra como claves y la cantidad como valor
        dic_final[primera_letra][palabra] = cantidad 
    return dic_final

resultado_indice = indice_palabras(resultado)
print(resultado_indice)

```

2.Conversió de tipus de dades.

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client:

```python
conversion=["1.3", "one", "1e10", "seven","3-1/2", ('2',1,1.4,'not-a-number'),[1,2,'3','3.4']]

#Creamos una función para convertir en float los elementos de una lista
def convert_data_type(lista):
    lista_float = []
    lista_s_float = []
    for data in lista:
        #Utilizamos el método try/except para manejar el error de valores al intentar adicionar los datos a la lista
        try:
            lista_float.append(float(data))
        except ValueError:
            lista_s_float.append(data)
        except TypeError:
            #En el caso de error de tipo de dato, llamamos a la función para alojar en dos sublistas, aquellos elementos sin convertir  
            sub_float, sub_s_float = convert_data_type(data)
            lista_float.extend(sub_float)
            lista_s_float.extend(sub_s_float)
    return lista_float, lista_s_float

# Ejecutar la función y mostrar los resultados
floats, not_floats = convert_data_type(conversion)
print("Elementos convertidos a float:", floats)
print("Elementos NO convertibles a float:", not_floats)
```

En este ejercicio se crea la función para convertir el tipo de dato contenido en una lista a "float". Para ello se crean dos listas vacias para alojar los números convertidos y los que no son viables de conversión. Y mediante la excepción try/except, se manejan los posibles errores de valor y tipo de dato. Se crean ademas, dos sublistas, para alojar elementos que no fueron considerados en el primer bucle, y llamamos a la función para que nuevamente se ejecute. Con el método "extend" combinamos los elementos de la primera iteración con la segunda, en cada una de las listas, según corresponda.

```python
import pandas as pd
df_partidos = pd.read_csv(r"C:\Users\jprom\Desktop\IT Academy\Python\Data Analytics\Sprint 9\historic partits.txt")

# Dividir la columna única, en tres, usando tab como separador
df_partidos_split = df_partidos[df_partidos.columns[0]].str.split('\t', expand=True)
df_partidos_split.columns = ["Local", "Resultado", "Visitante"]


df_partidos_split
```

```python
#Creamos una función que toma como argumento un df con los datos de los equipos (local, resultado y visitante), para calcular el total de goles por equipo

def total_goles(df):
    dic_total_goles_realizados = {}  # Inicializar el diccionario aquí
    for index, row in df.iterrows():
        local = row['Local']
        visitante = row['Visitante']
        resultado = row['Resultado']
        try:
            goles_local, goles_visitante = map(int, resultado.split('-'))
        except Exception:
            continue  # saltar si el resultado del formato es erróneo
        
        dic_total_goles_realizados[local] = dic_total_goles_realizados.get(local, 0) + goles_local
        dic_total_goles_realizados[visitante] = dic_total_goles_realizados.get(visitante, 0) + goles_visitante
        
      
    # Devolver el diccionario ordenado por cantidad de goles por equipo 
    return dict(sorted(dic_total_goles_realizados.items(), key=lambda item: item[1], reverse=True))
goles_por_equipo = total_goles(df_partidos_split)
goles_por_equipo
```

```python
def clasificacion(df):
    puntos = {}
    for index, row in df.iterrows():
        local = row['Local']
        visitante = row['Visitante']
        resultado = row['Resultado']
        try:
            goles_local, goles_visitante = map(int, resultado.split('-'))
        except Exception:
            continue  # saltar si el resultado del formato es erróneo

        # Inicializar equipos si no existen
        if local not in puntos:
            puntos[local] = 0
        if visitante not in puntos:
            puntos[visitante] = 0

        if goles_local > goles_visitante:
            puntos[local] += 3
        elif goles_local == goles_visitante:
            puntos[local] += 1
            puntos[visitante] += 1
        else:
            puntos[visitante] += 3
    # Ordenar por puntos descendente
    return dict(sorted(puntos.items(), key=lambda item: item[1], reverse=True))

clasificacion_global = clasificacion(df_partidos_split)
clasificacion_global
```
